# 2.2 Traditional Football Player Scouting

Football scouting traditionally involves scouts observing players through live matches, recorded games, and performance reports. Scouts assess characteristics such as technical ability, tactical awareness, physical attributes, decision-making, and positional suitability.

Although expert observation remains important, traditional scouting can be resource-intensive. # 2.3 Football Performance Analytics

Football performance analytics involves collecting and analyzing quantitative information about players and teams. Common player metrics include goals, assists, shots, passes, tackles, interceptions, duels, minutes played, and other event-based statistics.

However, raw statistics can be misleading when players have different amounts of playing time. For example, a player who scores 10 goals after playing 3,000 minutes should not necessarily be compared directly with a player who scores 8 goals after playing only 1,200 minutes. Consequently, researchers and analysts often use normalized measures such as statistics per 90 minutes.

Performance analytics can therefore provide a more standardized representation of player contribution. For a scouting system, these measurements can serve as input features for machine learning models that compare players according to their playing characteristics.

# 2.4 Machine Learning in Football Analytics

Machine learning provides methods for discovering patterns in datasets without requiring every relationship to be explicitly programmed. In sports analytics, supervised and unsupervised learning have been applied to areas including performance prediction, player valuation, tactical analysis, and recruitment.

For player scouting, unsupervised learning is particularly relevant because the objective may be to discover groups of players with similar characteristics rather than predict a predefined class. Clustering algorithms can group players according to statistical profiles, allowing players with comparable characteristics to be identified.

Potential techniques for this study include:

1. K-Means clustering
2. Gaussian Mixture Models
3. Hierarchical clustering
4. Principal Component Analysis (PCA)
5. Cosine similarity
6. Euclidean distance

These techniques can be used individually or in combination to construct a player similarity system.

# **2.5 Player Similarity Analysis**

Player similarity analysis attempts to identify players who exhibit comparable statistical profiles. Instead of comparing players using one statistic, machine-learning approaches can simultaneously consider multiple attributes.

For example, two attacking midfielders could be compared using:

1. Goals per 90
2. Assists per 90
3. Key passes
4. Progressive passes
5. Shots
6. Expected goals
7. Expected assists
8. Dribbling statistics

A similarity model can transform these characteristics into a multidimensional representation of each player. The distance between players in this feature space can then be used to determine which players have comparable profiles.

However, player similarity has an important limitation: statistical similarity does not necessarily mean tactical interchangeability. Two players may produce similar statistics while operating in different tactical systems or possessing different qualities that are difficult to quantify. Therefore, the proposed system should be viewed as a scouting-support tool rather than a replacement for professional scouting judgment.

# **2.6 Clustering Techniques for Player Identification**

## 2.6.1 K-Means Clustering

K-Means is an unsupervised machine-learning algorithm that divides observations into a predetermined number of clusters. In football analytics, it can be used to group players according to statistical characteristics.

For example, a dataset could potentially reveal clusters corresponding broadly to different attacking, midfield, or defensive profiles. The advantage of K-Means is its simplicity and computational efficiency. However, it requires the number of clusters to be specified in advance and assumes a particular structure for the clusters.

## 2.6.2 Gaussian Mixture Models

Gaussian Mixture Models (GMM) provide a probabilistic approach to clustering. Unlike K-Means, which assigns each observation to a single cluster, GMM can provide probabilities representing how strongly a player belongs to different clusters.

This can be useful in football because player roles are not always clearly separated. For example, a midfielder may possess characteristics associated with both creative and attacking midfielders. A probabilistic model can represent this overlap more effectively than a hard clustering approach.

## 2.6.3 Hierarchical Clustering

Hierarchical clustering creates a hierarchy of observations based on their similarity. This approach can be useful for exploring relationships between players without initially requiring a fixed number of clusters.

The resulting hierarchy can help researchers examine how players group together at different levels of similarity.

# 2.7 Dimensionality Reduction and PCA

Football datasets can contain many correlated variables. For example, goals, shots, expected goals, and attacking actions may contain overlapping information. Using a large number of correlated variables can make analysis more difficult.

Principal Component Analysis (PCA) can reduce the dimensionality of a dataset by transforming correlated variables into a smaller number of principal components while retaining as much variance as possible.

In the proposed system, PCA could be used to:

1. Reduce redundant information.
2. Visualize player profiles.
3. Identify major patterns in player performance.
4. Improve computational efficiency.
5. Support visualization of player clusters.

However, PCA components can be difficult to interpret because each component combines several original variables. Therefore, PCA should complement rather than replace meaningful football performance features.

# 2.8 Comparing Players Across Different Leagues

One of the major challenges in football player comparison is that players operate in different competitive environments. A player producing particular statistics in one league may not produce the same statistics when transferred to another league.

League strength can therefore be an important consideration when evaluating players. A scouting system that simply compares raw statistics could potentially favor players from environments where particular statistics are easier to accumulate.

The proposed research therefore considers incorporating a league-strength component into player evaluation. League strength could be estimated using measurable indicators such as:

1. International competition performance.
2. Club performance in continental competitions.
3. Historical league performance.
4. Average club strength.
5. Match outcomes between teams from different leagues.
6. Other available league-level performance indicators.

# 2.9 Player Market Value

Market value is another important factor in football recruitment. Clubs operate under financial constraints, meaning that identifying a statistically suitable player is not sufficient; the player's economic value may also influence whether recruitment is feasible.

Research into football economics has examined relationships between player characteristics, performance, age, position, and transfer values. Machine-learning regression techniques can potentially model these relationships.

Possible models include:

1. Linear Regression
2. Random Forest Regression
3. Gradient Boosting
4. XGBoost

The proposed system can use market value in two ways.

First, it can display the estimated or recorded market value of recommended players.

Second, market value can be used as a filtering or comparison criterion when identifying potential alternatives.

It is important to distinguish market value from actual transfer fee. A player's estimated market value does not necessarily equal the amount a club would pay to acquire the player because transfer negotiations can depend on factors such as contract length, club circumstances, demand, and negotiation conditions.

# 2.10 Existing Approaches and Research Gap

Existing football analytics approaches demonstrate that statistical and machine-learning techniques can support player evaluation and recruitment. However, several challenges remain.

First, player similarity systems can focus heavily on performance statistics without adequately accounting for differences between leagues.

Second, systems that identify statistically similar players may not incorporate economic considerations such as market value.

Third, market-value models may focus primarily on prediction rather than using valuation as part of a broader player-recruitment process.

This creates an opportunity for an integrated system combining:

Player performance + player similarity + league strength + market value

The proposed research aims to investigate how these components can be combined into a single scouting-support system. The system would allow a user to select a reference player and identify other players with comparable statistical profiles, while providing additional context regarding the player's league and market value.

Build basic player performance totals

We'll calculate each club's:

Then we'll aggregate those club performances to the league level. This gives us a stronger foundation for the league-strength feature.

```python
home_clubs = league_games_2025[
    [
        "competition_id",
        "home_club_id",
        "home_club_goals",
        "away_club_goals",
        "home_points"
    ]
].copy()

home_clubs = home_clubs.rename(
    columns={
        "home_club_id": "club_id",
        "home_club_goals": "goals_for",
        "away_club_goals": "goals_against",
        "home_points": "points"
    }
)

away_clubs = league_games_2025[
    [
        "competition_id",
        "away_club_id",
        "away_club_goals",
        "home_club_goals",
        "away_points"
    ]
].copy()

away_clubs = away_clubs.rename(
    columns={
        "away_club_id": "club_id",
        "away_club_goals": "goals_for",
        "home_club_goals": "goals_against",
        "away_points": "points"
    }
)

club_matches = pd.concat(
    [home_clubs, away_clubs],
    ignore_index=True
)

club_performance = (
    club_matches
    .groupby(
        ["competition_id", "club_id"],
        as_index=False
    )
    .agg(
        matches=("club_id", "count"),
        points=("points", "sum"),
        goals_for=("goals_for", "sum"),
        goals_against=("goals_against", "sum")
    )
)

club_performance["points_per_game"] = (
    club_performance["points"] /
    club_performance["matches"]
)

club_performance["goal_difference"] = (
    club_performance["goals_for"] -
    club_performance["goals_against"]
)

club_performance["goal_difference_per_game"] = (
    club_performance["goal_difference"] /
    club_performance["matches"]
)

print("Club performance calculated")
print("===========================")
print("Club-league records:", len(club_performance))
print("Unique clubs:", club_performance["club_id"].nunique())

display(club_performance.head(10))
```

Aggregate club performance to the league level

```python
league_club_strength = (
    club_performance
    .groupby("competition_id")
    .agg(
        clubs=("club_id", "nunique"),
        avg_points_per_game=("points_per_game", "mean"),
        avg_goal_difference_per_game=("goal_difference_per_game", "mean"),
        avg_goals_for_per_game=(
            "goals_for",
            lambda x: x.sum() / club_performance.loc[x.index, "matches"].sum()
        ),
        avg_goals_against_per_game=(
            "goals_against",
            lambda x: x.sum() / club_performance.loc[x.index, "matches"].sum()
        )
    )
    .reset_index()
)

league_club_strength = league_club_strength.merge(
    competitions[
        [
            "competition_id",
            "name",
            "country_name"
        ]
    ],
    on="competition_id",
    how="left"
)

print("League club-strength summary")
print("============================")
print("Leagues:", len(league_club_strength))

display(
    league_club_strength[
        [
            "competition_id",
            "name",
            "country_name",
            "clubs",
            "avg_points_per_game",
            "avg_goal_difference_per_game",
            "avg_goals_for_per_game",
            "avg_goals_against_per_game"
        ]
    ].head(15)
)
```

What this gives us

Instead of looking at something like:

"Brazil has a higher home goal difference"

we'll now have measures such as:

Average points per game → how well clubs perform
Average goal difference per game → competitive performance
Goals scored/conceded per game → attacking and defensive environment

These are much more useful inputs for constructing our eventual league_strength_score.

Find cross-league matches

We'll inspect 2025 international-cup games and identify matches between clubs whose domestic leagues are different.

```python
international_games_2025 = con.execute("""
    SELECT
        g.game_id,
        g.competition_id,
        c.name AS competition_name,
        g.date,
        g.home_club_id,
        g.away_club_id,
        g.home_club_goals,
        g.away_club_goals
    FROM games g
    INNER JOIN competitions c
        ON g.competition_id = c.competition_id
    WHERE
        c.type = 'international_cup'
        AND EXTRACT(YEAR FROM g.date) = 2025
""").fetchdf()

print("2025 international-cup games:", len(international_games_2025))
print("International competitions:",
      international_games_2025["competition_id"].nunique())

display(international_games_2025.head(10))
```

This is the next building block for a cross-league strength model. If the data contains enough cross-league matches, we can use those results to estimate relative league strength rather than assigning subjective league ratings.

Map clubs to their domestic leagues

```python
club_leagues = con.execute("""
    SELECT
        club_id,
        name AS club_name,
        domestic_competition_id
    FROM clubs
""").fetchdf()

club_leagues = club_leagues.merge(
    competitions[
        [
            "competition_id",
            "name",
            "country_name",
            "type",
            "sub_type"
        ]
    ],
    left_on="domestic_competition_id",
    right_on="competition_id",
    how="left",
    suffixes=("_club", "_league")
)

club_leagues = club_leagues.rename(
    columns={
        "name_league": "league_name",
        "country_name": "league_country"
    }
)

print("Club-league mapping")
print("===================")
print("Clubs:", len(club_leagues))

display(
    club_leagues[
        [
            "club_id",
            "club_name",
            "domestic_competition_id",
            "league_name",
            "league_country"
        ]
    ].head(15)
)
```
