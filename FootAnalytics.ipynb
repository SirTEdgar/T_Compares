# T_Search
The football industry increasingly relies on data and machine learning to support player recruitment and scouting decisions.

Traditional scouting often depends heavily on subjective assessments, limited observations, and the experience of individual scouts. With the growing availability of football performance and market data, machine learning provides an opportunity to make the scouting process more systematic, data-driven, and efficient.

This project proposes the development of a Machine Learning Player Scouting Tool that identifies football players who are statistically similar to a selected reference player.

The system will analyse attributes such as player position, age, performance statistics, playing characteristics, and market value to discover potential alternatives or emerging talents.

In addition to player similarity, the tool will consider the strength and quality of the leagues in which players compete. This is important because statistics achieved in different leagues may not be directly comparable. The system will therefore incorporate league-level factors when interpreting player performances.

It will also analyse player market values to help identify players who provide similar characteristics at different financial levels.

The final tool will allow a user to select a player and receive a list of similar players, together with information about their performance, league strength, and estimated market value. By combining machine learning, player similarity analysis, league evaluation, and market-value analysis, the project aims to provide a practical data-driven approach to football scouting and recruitment.
